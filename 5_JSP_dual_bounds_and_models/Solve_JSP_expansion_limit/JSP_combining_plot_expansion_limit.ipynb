```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import math

# ==========================================
# ⚙️ CONFIGURATION: PATHS
# ==========================================

# 1. Path to the SINGLE BOUND "Profiles" folder
DIR_SINGLE = r"C:\Users\ACER\Desktop\Code\0_Thesis_implementation\2_DIDP_custom_search_guidance_local\Thesis_modified_DIDP\5_JSP_dual_bounds_and_models\Solve_JSP_expansion_limit\Single_dual_bound\JSP_single_dual_bound_Profiles_Parsed"

# 2. Path to the EA BOUND "Results" folder
DIR_EA = r"C:\Users\ACER\Desktop\Code\0_Thesis_implementation\2_DIDP_custom_search_guidance_local\Thesis_modified_DIDP\5_JSP_dual_bounds_and_models\Solve_JSP_expansion_limit\EA_dual_bounds\JSP_EA_dual_bound_Parsing_Expansion_Limit_Results"

# Output for the combined plot
OUTPUT_PLOT_PATH = "Combined_Convergence_Comparison.png"

# CSV Filenames (Standard names from previous scripts)
FILE_SINGLE = "JSP_single_dual_bound_Average_gaps_vs_Expansion_limit.csv"
FILE_EA = "JSP_Average_Gaps_vs_Expansion.csv"

# ==========================================
# 📈 PLOTTING FUNCTION
# ==========================================
def create_synchronized_plot():
    print("🔹 Generating Synchronized Comparison Plot...")

    # --- 1. Load Data ---
    path_single = os.path.join(DIR_SINGLE, FILE_SINGLE)
    path_ea = os.path.join(DIR_EA, FILE_EA)

    if not os.path.exists(path_single):
        print(f"❌ Missing Single Bound CSV: {path_single}")
        return
    if not os.path.exists(path_ea):
        print(f"❌ Missing EA Bound CSV: {path_ea}")
        return

    df_single = pd.read_csv(path_single)
    df_ea = pd.read_csv(path_ea)

    # --- 2. Setup Plot ---
    plt.figure(figsize=(12, 7))

    # Plot Single Bound (Red Dashed)
    plt.plot(
        df_single['Node_Count'], 
        df_single['Average_Gap_Percent'], 
        label='CABS with h = Single dual bound', 
        color='red', 
        linestyle='--', 
        linewidth=2,
        alpha=0.8
    )

    # Plot EA Bound (Blue Solid)
    plt.plot(
        df_ea['Node_Count'], 
        df_ea['Average_Gap_Percent'], 
        label='CABS with h= EA-constructed dual bound', 
        color='blue', 
        linestyle='-', 
        linewidth=2
    )

    # --- 3. Format Y-Axis (Step 5) ---
    # Find global max to set upper limit
    max_single = df_single['Average_Gap_Percent'].max()
    max_ea = df_ea['Average_Gap_Percent'].max()
    global_max = max(max_single, max_ea)

    if pd.isna(global_max): global_max = 100

    upper_limit = 400
    step = 100
    
    # Set ticks: 0, 5, 10...
    plt.yticks(np.arange(0, upper_limit + step, step))
    plt.ylim(0, upper_limit)

    # --- 4. Labels & Style ---
    plt.xlabel('Nodes Expanded', fontsize=12)
    plt.ylabel('Average Optimality Gap (%)', fontsize=12)
    plt.title('Convergence Profile JSP', fontsize=14)
    plt.legend(fontsize=12)
    plt.grid(True, linestyle=':', alpha=0.6)

    # --- 5. Save ---
    plt.tight_layout()
    plt.savefig(OUTPUT_PLOT_PATH, dpi=300)
    print(f"✅ Comparison Plot Saved: {os.path.abspath(OUTPUT_PLOT_PATH)}")
    plt.show()

if __name__ == "__main__":
    create_synchronized_plot()
```
