```python
import sys
import cplex
from docplex.mp.model import Model

def test_cplex_setup():
    print(f"Python Version: {sys.version.split()[0]}")
    
    # 1. Check CPLEX Engine
    try:
        c = cplex.Cplex()
        print(f"✅ CPLEX Engine Found: Version {c.get_version()}")
    except Exception as e:
        print(f"❌ CPLEX Engine Error: {e}")
        return

    # 2. Check Docplex Wrapper (Simple Optimization)
    print("\nRunning simple optimization test...")
    try:
        mdl = Model(name='test_model')
        x = mdl.continuous_var(name='x')
        y = mdl.continuous_var(name='y')
        
        # Constraint: x + y <= 10
        mdl.add_constraint(x + y <= 10)
        # Objective: Maximize x + y
        mdl.maximize(x + y)
        
        solution = mdl.solve()
        
        if solution:
            print(f"✅ Optimization Success!")
            print(f"   Objective Value: {solution.objective_value} (Expected: 10.0)")
            print("   Your setup is PERFECT.")
        else:
            print("⚠️ Model solved but no solution found.")
            
    except Exception as e:
        print(f"❌ Docplex Error: {e}")

if __name__ == "__main__":
    test_cplex_setup()
```
