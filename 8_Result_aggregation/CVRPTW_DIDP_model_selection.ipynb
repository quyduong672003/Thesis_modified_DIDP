```python
import pandas as pd

# --- CONFIGURATION ---
file_1t = 'CVRPTW_1T_single_dual_bound_selected_results_1800s_lim.csv'
file_2t = 'CVRPTW_2T_single_dual_bound_selected_results_1800s_lim.csv'
output_excel = 'CVRPTW_Model_Comparison.xlsx'

# --- LOAD DATA ---
df1 = pd.read_csv(file_1t)
df2 = pd.read_csv(file_2t)

# --- PREPARE DATA ---
# Select and rename columns for 1T Model
df1_clean = df1[['Instance', 'Best Known Cost', 'Cost', 'Running Time (s)', 'Nodes Expanded']].copy()
df1_clean.rename(columns={
    'Best Known Cost': 'Best Known Solution',
    'Cost': 'CVRPTW-1T Solution',
    'Running Time (s)': 'CVRPTW-1T Time',
    'Nodes Expanded': 'CVRPTW-1T Expanded Nodes'
}, inplace=True)

# Select and rename columns for 2T Model
df2_clean = df2[['Instance', 'Cost', 'Running Time (s)', 'Nodes Expanded']].copy()
df2_clean.rename(columns={
    'Cost': 'CVRPTW-2T Solution',
    'Running Time (s)': 'CVRPTW-2T Time',
    'Nodes Expanded': 'CVRPTW-2T Expanded Nodes'
}, inplace=True)

# Merge the two datasets
merged = pd.merge(df1_clean, df2_clean, on='Instance', how='outer')

# --- CALCULATIONS ---
# Calculate Optimality Gaps: |Val - Best| / Best * 100
merged['CVRPTW-1T Optimality Gap'] = (
    abs(merged['CVRPTW-1T Solution'] - merged['Best Known Solution']) / merged['Best Known Solution'] * 100
)
merged['CVRPTW-2T Optimality Gap'] = (
    abs(merged['CVRPTW-2T Solution'] - merged['Best Known Solution']) / merged['Best Known Solution'] * 100
)

# Reorder columns for the final Detailed Table
cols_order = [
    'Instance', 'Best Known Solution',
    'CVRPTW-1T Solution', 'CVRPTW-1T Optimality Gap', 'CVRPTW-1T Time', 'CVRPTW-1T Expanded Nodes',
    'CVRPTW-2T Solution', 'CVRPTW-2T Optimality Gap', 'CVRPTW-2T Time', 'CVRPTW-2T Expanded Nodes'
]
detailed_df = merged[cols_order].copy()

# --- COMPUTE AVERAGES ---
# Calculate means for numeric columns
avgs = detailed_df.mean(numeric_only=True)

# Create the Average Row dictionary
avg_row = {
    'Instance': 'Average',
    'Best Known Solution': avgs['Best Known Solution'],
    'CVRPTW-1T Solution': avgs['CVRPTW-1T Solution'],
    'CVRPTW-1T Optimality Gap': avgs['CVRPTW-1T Optimality Gap'],
    'CVRPTW-1T Time': avgs['CVRPTW-1T Time'],
    'CVRPTW-1T Expanded Nodes': round(avgs['CVRPTW-1T Expanded Nodes']), # Rounding here as well for consistency
    'CVRPTW-2T Solution': avgs['CVRPTW-2T Solution'],
    'CVRPTW-2T Optimality Gap': avgs['CVRPTW-2T Optimality Gap'],
    'CVRPTW-2T Time': avgs['CVRPTW-2T Time'],
    'CVRPTW-2T Expanded Nodes': round(avgs['CVRPTW-2T Expanded Nodes']) # Rounding here as well
}

# Append Average Row to Detailed DataFrame
detailed_df_with_avg = pd.concat([detailed_df, pd.DataFrame([avg_row])], ignore_index=True)

# --- CREATE SUMMARY SHEET ---
# Create a small summary table with averages, min/max gap, and rounded nodes
summary_data = [
    {
        'Model': 'CVRPTW-1T',
        'Average Optimality Gap (%)': avgs['CVRPTW-1T Optimality Gap'],
        'Smallest Gap (%)': detailed_df['CVRPTW-1T Optimality Gap'].min(), # Min Gap
        'Largest Gap (%)': detailed_df['CVRPTW-1T Optimality Gap'].max(), # Max Gap
        'Average Time (s)': avgs['CVRPTW-1T Time'],
        'Average Expanded Nodes': round(avgs['CVRPTW-1T Expanded Nodes']) # Rounded
    },
    {
        'Model': 'CVRPTW-2T',
        'Average Optimality Gap (%)': avgs['CVRPTW-2T Optimality Gap'],
        'Smallest Gap (%)': detailed_df['CVRPTW-2T Optimality Gap'].min(), # Min Gap
        'Largest Gap (%)': detailed_df['CVRPTW-2T Optimality Gap'].max(), # Max Gap
        'Average Time (s)': avgs['CVRPTW-2T Time'],
        'Average Expanded Nodes': round(avgs['CVRPTW-2T Expanded Nodes']) # Rounded
    }
]
summary_df = pd.DataFrame(summary_data)

# --- SAVE TO EXCEL ---
with pd.ExcelWriter(output_excel, engine='xlsxwriter') as writer:
    # Sheet 1: Detailed Comparison (with Average row)
    detailed_df_with_avg.to_excel(writer, sheet_name='Detailed Comparison', index=False)
    
    # Sheet 2: Model Summary (Just averages)
    summary_df.to_excel(writer, sheet_name='Model Summary', index=False)
    
    # Optional: Adjust column widths
    worksheet1 = writer.sheets['Detailed Comparison']
    worksheet1.set_column(0, 0, 15)
    worksheet1.set_column(1, 10, 20)
    
    worksheet2 = writer.sheets['Model Summary']
    worksheet2.set_column(0, 5, 25)

print(f"Successfully created {output_excel}")
print("\n--- Model Summary Preview ---")
print(summary_df.to_string(index=False))
```
